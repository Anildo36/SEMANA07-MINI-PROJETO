```python
# importando Biblioteca
import pandas as pd
```

```python
# Criando DataFrame
# Ao abrir Os arquivos vir que são separados por ponto e vírgula e não somente por vírgula
# utilizei o sep=";" e se tiver acentos o encording="utf-8" vai garantir que sejam lidos corretamente.

df = pd.read_csv("Base Varejo.csv",sep=";",encoding="utf-8")

df
```

```python
# Verificando como cada Dado analisado se encontra para as devidas correções.
df.dtypes
```

# Analisando a Tabela

```python
# Pecorre toda minha tabela e verifica onde há valores nulos e vazios , e me mostra a quantidade os somando

df.isnull().sum()
```

```python
# Verificando quantas linhas duplicadas há no arquivo e me mostra a quantidade

print("Linhas_Duplicadas" , df.duplicated().sum)
```

```python
# Verificando se há produtos com categoria vazia, onde ela tava usando o #N/D como um nulo desfarçado

print("Produtos_vazios (#N/D):", (df["PR_CAT"]=="#N/D").sum())
```

```python
# Verificando se as datas estão corretas utilizando o datetime que vai converter texto em data  caso esteja
# E utilizando o isna() para contar as falhas,  e o errors="coerce" para transformar um vaalor invalido em vazio

df_data_correta = pd.to_datetime(df["DATA"], format= "%d/%m/%Y",errors="coerce")

df_data_correta.isna().sum()
```

# Limpeza de Dados

```python
# Criando uma cópia dos Dados para uma futura comparação mantendo o Dado original intacto, e fazer a limpeza toda nessa cópia
# Uma vez que já se sabe o que vai ser Limpo.

df_Limpeza_Dados = df.copy()
df
```

```python
# Tratando Colunas Vazias eliminando elas

df = df.drop(columns=["Unnamed: 10", "Unnamed: 11", "Unnamed: 12", "Unnamed: 13"])

df
```

```python
# Tratando Produtos Nulos trabalhando encima da cópia utilizando o replace para trocar o texto #N/D por Produto Vazio,
#  e o isnull pra confirmar se tudo ta 100% sem nulos e conta-los com sum()

df_Limpeza_Dados["PR_CAT"] = df_Limpeza_Dados["PR_CAT"].replace("#N/D", "Sem Categoria")

df_Limpeza_Dados.isnull().sum()
```

```python
# Eliminando as Duplicadas utilizando o len que vai me contar quantas linhas tem duplicadas antes de ser removida guardando pra ser comparada depois 
# Enquanto isso o drop_duplicate vai limpando as linhas repetidas mantendo a ocorrência inicial ou seja a primeira linha de cada uma 
# E foi feita uma subtração de quantas linhas sairam em ( qtd_linhas_antes-len(df_limpeza_Dados) )

qtd_linhas_antes  = len(df_Limpeza_Dados)
print("Linhas_Iniciais",len(df_Limpeza_Dados))

df_Limpeza_Dados = df_Limpeza_Dados.drop_duplicates()
print("Linhas_Duplicadas: ",qtd_linhas_antes-len(df_Limpeza_Dados))
```
